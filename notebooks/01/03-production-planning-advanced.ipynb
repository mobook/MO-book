- [Sets](https://pyomo.readthedocs.io/en/6.8.0/pyomo_modeling_components/Sets.html)

These components enable the use of indexed variables and constraints. The combination of sets and indices is essential to building scalable and maintainable models for more complex applications.

We will begin this analysis by examining the problem data sets to identify the underlying problem structure.

## Mathematical model

By rearranging the problem data into straightforward tables, the structure of the production planning problem becomes evident. We can identify a set of products, a set of resources, and a collection of parameters that specify the processes for transforming resources into products. Compared to the previous notebook, these abstractions allow us to create mathematical models that can adapt and scale with the supplied data.

Let $\cal{P}$ and $\cal{R}$ be the set of products and resources, respectively, and let $p$ and $r$ be representative elements of those sets. We use indexed decision variables $x_r$ to denote the amount of resource $r$ that is consumed in production, and $y_p$ to denote the amount of product $p$ produced.

The problem data provides attributes that constrain the decision variables. For example, the decision variables all have lower bounds of zero, and some have upper bounds. We represent these as

$$
\begin{aligned}
    0 \leq x_r \leq b_r & & \forall r\in\cal{R} \\
    0 \leq y_p \leq b_p & & \forall p\in\cal{P} \\
\end{aligned}
$$

where the upper bounds, $b_r$ and $b_p$, come from the tables of attributes. For cases where the upper bounds don't apply, we can either insert bounds larger than would ever be encountered, or, when we translate this model to Pyomo, designate a special value that causes the bound to be ignored.

The objective is given as before,

$$
\begin{aligned}
    \text{profit} & = \text{revenue} - \text{cost} \\
\end{aligned}
$$

but now the expressions for revenue and cost are

$$
\begin{aligned}
    \text{revenue} & = \sum_{p\in\cal{P}} c_p y_p  \\
    \text{cost} & = \sum_{r\in\cal{R}} c_r x_r \\
\end{aligned}
$$

where $c_r$ and $c_p$ are parameters specifying the price for resources and products. The bounds on available resources can be written as

$$
\begin{aligned}
    \sum_{p\in\cal{P}} a_{r, p} y_p & \leq x_r & \forall r\in\cal{R}
\end{aligned}
$$

Putting these pieces together, we have the following model for the production planning problem.

$$
\begin{align}
\max \quad & \text{profit} = \sum_{p\in\cal{P}} c_p y_p - \sum_{r\in\cal{R}} c_r x_r \\
\text{such that} \quad & \sum_{p\in\cal{P}} a_{r, p} y_p  \leq x_r & \forall r\in\cal{R} \nonumber \\
 &   0 \leq x_r \leq b_r & \forall r\in\cal{R} \nonumber  \\
 &   0 \leq y_p \leq b_p & \forall p\in\cal{P} \nonumber  \\
\end{align}
$$

Compared to the previous notebook, when formulated this way, the model can be applied to any problem with the same structure, regardless of the number of products or resources. This flexibility is possible due to the use of sets to describe the products and resources for a particular problem, indices to refer to elements of those sets, and data tables that hold the relevant parameter values.

Generalizing mathematical models in this fashion is a common feature of many data science applications. Next we will see how this type of generalization is facilitated in Pyomo.


## The production model in Pyomo

As before, we begin the construction of a Pyomo model by creating a `ConcreteModel`.

We first use component constructors throughout. After solving the model, we show the corresponding decorator version for indexed parameters and constraints, as well as expressions and the objective.

````python
model = pyo.ConcreteModel()
````

In mathematical optimization and modeling, a set serves as an indexed collection of elements that allows you to define variables, constraints, and other model components in a generalized way. Pyomo's `Set()` component serves the same purpose: it is used for defining index sets over which variables, parameters, constraints, or objectives can be defined.

We use the Pyomo `Set()` to construct sets corresponding to the products and resources. Each set is initialized with the dictionary keys for the relevant attribute tables. These will later become indices for parameters, decision variables, and constraints.


````python
model.PRODUCTS = pyo.Set(initialize=products.keys())
model.RESOURCES = pyo.Set(initialize=resources.keys())
````

The next step is to introduce parameters that will be used in the constraints and objective. These are indexed by products, resources, or both. Each `pyo.Param()` constructor receives the relevant index sets and an `initialize` function. Pyomo calls that function with the model followed by the index values to obtain each parameter value from the problem data. This forms the interface between the data and the Pyomo model.

This step of declaring Pyomo `Param` objects is often omitted in Pyomo applications. In doing so, the modeler is making a choice to embed the external data representation directly into objectives and constraints of the problem. This can be effective, it does keep the code shorter, and may remove some computational overhead. However, it also blurs the boundary between the data representation and the model statements. Any change in data representation may require editing every place where that data is used in the model. By defining model parameters with `Param()`, the interface to the data representation is located in one clearly defined portion of a larger model, thereby significantly improving the long-term maintainability of models. This concern may be overkill in small models like we have here, but is a key consideration when constructing more complex applications.

Note: The domain for the bounds is set to `Any` because some of them will take value `None`. Pyomo will omit a lower or upper bound that has a value of `None`, so this is a way to keep the logic simple.

````python
# parameters for bounds
def demand_rule(m, product):
    return products[product]["demand"]


model.demand = pyo.Param(model.PRODUCTS, initialize=demand_rule, domain=pyo.Any)


def available_rule(m, resource):
    return resources[resource]["available"]


model.available = pyo.Param(model.RESOURCES, initialize=available_rule, domain=pyo.Any)


# parameters with price coefficients
def cp_rule(m, product):
    return products[product]["price"]


model.cp = pyo.Param(model.PRODUCTS, initialize=cp_rule)


def cr_rule(m, resource):
    return resources[resource]["price"]


model.cr = pyo.Param(model.RESOURCES, initialize=cr_rule)


# process parameters: a[r,p]
def a_rule(m, resource, product):
    return processes[product][resource]


model.a = pyo.Param(model.RESOURCES, model.PRODUCTS, initialize=a_rule)
````

The decision variables, $x$ and $y$, are indexed by the set of resources and products, respectively. The indexing is specified by passing the relevant sets as the first arguments to Pyomo `Var()`. In addition to indexing, it is always good practice to specify any known and fixed bounds on the variables. This is done by specifying a function (in Pyomo parlance sometimes called a rule) that returns the bound for a given index. Here we use a Python lambda function with two arguments, model and an index referring to a member of a set, to return a tuple with the lower and upper bound.


````python
model.x = pyo.Var(
    model.RESOURCES, bounds=lambda model, resource: (0, model.available[resource])
)
model.y = pyo.Var(
    model.PRODUCTS, bounds=lambda model, product: (0, model.demand[product])
)
````

The objective is expressed with Pyomo `quicksum` which accepts a Python generator to successive terms in the sum. Here we use the parameters $c_p$ and $c_r$ that appear in the mathematical version of the model, and which were declared above in the Pyomo version of the model.

As in the previous notebook, named `Expression` components hold revenue and cost, which are then used in the objective.

````python
model.revenue = pyo.Expression(
    expr=pyo.quicksum(model.cp[product] * model.y[product] for product in model.PRODUCTS)
)
model.cost = pyo.Expression(
    expr=pyo.quicksum(model.cr[resource] * model.x[resource] for resource in model.RESOURCES)
)

# create objective
model.profit = pyo.Objective(expr=model.revenue - model.cost, sense=pyo.maximize)
````

The `pyo.Constraint()` constructor accepts index sets and a `rule` function. For each resource, Pyomo calls the rule with the model and the resource index to construct the associated constraint. This creates one resource-use constraint per member of `model.RESOURCES`.

````python
# create indexed constraint
def materials_used_rule(m, resource):
    return (
        pyo.quicksum(m.a[resource, product] * m.y[product] for product in m.PRODUCTS)
        <= m.x[resource]
    )


model.materials_used = pyo.Constraint(model.RESOURCES, rule=materials_used_rule)
````

The final step is to solve the model and report the solution. Here we create a simple report using `pyo.value()` to access values of the decision variables, and using the model sets to construct iterators to report the value of indexed variables.


````python
model.pprint()

# solve
results = SOLVER.solve(model)
pyo.assert_optimal_termination(results)

# create a solution report
print(f"Profit = {pyo.value(model.profit)}")

print("\nProduction Report")
for product in model.PRODUCTS:
    print(f" {product}  produced =  {pyo.value(model.y[product])}")

print("\nResource Report")
for resource in model.RESOURCES:
    print(f" {resource} consumed = {pyo.value(model.x[resource])}")
````

## The same indexed model with decorators

The [previous notebook](02-production-planning-basic.ipynb#the-same-model-with-decorators) explains how a decorator creates a component using the name and rule of the function below it. The same pattern applies to indexed components. The index sets go in the decorator, and the function receives the model followed by the indices in the same order.

For example, the constructor version used `model.a = pyo.Param(model.RESOURCES, model.PRODUCTS, initialize=a_rule)`. Its decorator counterpart is:

```python
@decorated_model.Param(decorated_model.RESOURCES, decorated_model.PRODUCTS)
def a(m, resource, product):
    return processes[product][resource]
```

Pyomo calls this function once for each resource-product pair to initialize `decorated_model.a[resource, product]`. For an indexed constraint, the returned object is a relationship involving decision variables. For a parameter, it is the corresponding data value.

The following cell builds a separate model from the same dictionaries. It uses decorators for the parameters, revenue and cost expressions, objective and resource-use constraints.

````python
decorated_model = pyo.ConcreteModel()
decorated_model.PRODUCTS = pyo.Set(initialize=products.keys())
decorated_model.RESOURCES = pyo.Set(initialize=resources.keys())


@decorated_model.Param(decorated_model.PRODUCTS, domain=pyo.Any)
def demand(m, product):
    return products[product]["demand"]


@decorated_model.Param(decorated_model.RESOURCES, domain=pyo.Any)
def available(m, resource):
    return resources[resource]["available"]


@decorated_model.Param(decorated_model.PRODUCTS)
def cp(m, product):
    return products[product]["price"]


@decorated_model.Param(decorated_model.RESOURCES)
def cr(m, resource):
    return resources[resource]["price"]


@decorated_model.Param(decorated_model.RESOURCES, decorated_model.PRODUCTS)
def a(m, resource, product):
    return processes[product][resource]


decorated_model.x = pyo.Var(
    decorated_model.RESOURCES, bounds=lambda m, resource: (0, m.available[resource])
)
decorated_model.y = pyo.Var(
    decorated_model.PRODUCTS, bounds=lambda m, product: (0, m.demand[product])
)


@decorated_model.Expression()
def revenue(m):
    return pyo.quicksum(m.cp[product] * m.y[product] for product in m.PRODUCTS)


@decorated_model.Expression()
def cost(m):
    return pyo.quicksum(m.cr[resource] * m.x[resource] for resource in m.RESOURCES)


@decorated_model.Objective(sense=pyo.maximize)
def profit(m):
    return m.revenue - m.cost


@decorated_model.Constraint(decorated_model.RESOURCES)
def materials_used(m, resource):
    return (
        pyo.quicksum(m.a[resource, product] * m.y[product] for product in m.PRODUCTS)
        <= m.x[resource]
    )
````

We solve the decorator model and compare its variables, revenue, cost and profit with the constructor model. For these data the two styles give the same solution; the mathematical formulation has not changed.

````python
decorated_results = SOLVER.solve(decorated_model)
pyo.assert_optimal_termination(decorated_results)


def solution_values(m):
    return {
        **{f"y[{p}]": pyo.value(m.y[p]) for p in m.PRODUCTS},
        **{f"x[{r}]": pyo.value(m.x[r]) for r in m.RESOURCES},
        "revenue": pyo.value(m.revenue),
        "cost": pyo.value(m.cost),
        "profit": pyo.value(m.profit),
    }


import pandas as pd

comparison = pd.DataFrame(
    {"Constructors": solution_values(model), "Decorators": solution_values(decorated_model)}
)
assert (abs(comparison["Constructors"] - comparison["Decorators"]) < 1e-7).all()
display(comparison)
````

## For Python experts: Creating subclasses of `ConcreteModel`

Some readers of these notebooks may be more experienced Python developers who wish to apply Pyomo in more specialized, data driven applications. The following cell shows how the Pyomo `ConcreteModel()` class can be extended by subclassing to create specialized model classes. Here we create a subclass called `ProductionModel` that accepts a particular representation of the problem data to produce a production model object. The production model object inherits all of the methods associated with any `ConcreteModel`, such as `.display()` and `.pprint()`. The subclass defines its own `.solve()` and `.report()` methods.


````python
class ProductionModel(pyo.ConcreteModel):
    """
    A class representing a production model using Pyomo.
    """

    def __init__(self, products, resources, processes):
        """
        Initialize ProductionModel as an instance of a ConcreteModel.

        :param products: A dictionary containing product information.
        :param resources: A dictionary containing resource information.
        :param processes: A dictionary containing process information.
        """
        super().__init__("Production Model")

        # save data in the model instance
        self.products = products
        self.resources = resources
        self.processes = processes

        # flags to monitor model construction and solution status
        self.built = False
        self.solved = False

    def build_model(self):
        """
        Build the optimization model.
        """
        # access the model
        model = self.model()

        # create sets to index variables and constraints
        model.PRODUCTS = self.products.keys()
        model.RESOURCES = self.resources.keys()

        # decision variables
        model.x = pyo.Var(
            model.RESOURCES,
            bounds=lambda model, resource: (0, self.resources[resource]["available"]),
        )
        model.y = pyo.Var(
            model.PRODUCTS,
            bounds=lambda model, product: (0, self.products[product]["demand"]),
        )

        # use expressions to break up complex objectives
        model.revenue = pyo.quicksum(
            self.products[product]["price"] * model.y[product]
            for product in model.products
        )
        model.cost = pyo.quicksum(
            self.resources[resource]["price"] * model.x[resource]
            for resource in model.resources
        )

        # create objective
        @model.Objective(sense=pyo.maximize)
        def profit(model):
            return model.revenue - model.cost

        # create indexed constraint
        @model.Constraint(model.RESOURCES)
        def materials_used(model, resource):
            return (
                pyo.quicksum(
                    self.processes[product][resource] * model.y[product]
                    for product in model.PRODUCTS
                )
                <= model.x[resource]
            )

        self.built = True

    def solve(self, solver=SOLVER):
        """
        Build the model, if necessary, then solve the optimization model.
        """
        if not self.built:
            self.build_model()
        self.solved = False
        results = solver.solve(self)
        pyo.assert_optimal_termination(results)
        self.solved = True
        return results

    def report(self):
        """
        Solve, if necessary, then report the model solution.
        """
        if not self.solved:
            self.solve(SOLVER)
        print(f"Profit = {pyo.value(self.profit)}")
        print("\nProduction Report")
        for product in self.PRODUCTS:
            print(f" {product}  produced =  {pyo.value(self.y[product])}")
        print("\nResource Report")
        for resource in self.RESOURCES:
            print(f" {resource} consumed = {pyo.value(self.x[resource])}")


m = ProductionModel(products, resources, processes)
m.report()
````
